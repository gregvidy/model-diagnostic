# Duration since First Time Trnx

```python
df_mock.rename({"Country Code": "CountryCode"}, axis=1, inplace=True)
```

```python
df_mock_first_time_duration = generate_rolling_features(
    df_mock,
    datetime_col="Transaction Datetime",
    key_col="Transaction Serial No",
    features_config=first_time_trnx_duration_config,
)
```

```python
df_mock_first_time_duration = df_mock_first_time_duration.sort_values(
    by=["PANNumber", "Transaction Datetime"]
)
for groupby_col in ["MCC", "CountryCode"]:
    df_mock_first_time_duration[f"first_txn_time_by_{groupby_col}"] = (
        df_mock_first_time_duration.groupby(["PANNumber", groupby_col])[
            "Transaction Datetime"
        ].transform("first")
    )

    amount_col = f"DurationSince_FirstTrnx_to_Current_{groupby_col}"
    df_mock_first_time_duration[amount_col] = (
        df_mock_first_time_duration["Transaction Datetime"]
        - df_mock_first_time_duration[f"first_txn_time_by_{groupby_col}"]
    ).dt.total_seconds() / 60

    df_mock_first_time_duration.drop(
        [f"first_txn_time_by_{groupby_col}"], axis=1, inplace=True
    )
```

```python
selected_cols = [
    col for col in df_mock_first_time_duration.columns if "DurationSince" in col
]
```

```python
dynamic_high_risk_mcc_config = {
    "groupby": "MCC",
    "fraud_label_col": "Confirmed",
    "agg_func": "sum",
    "months_period": 12,  # e.g. last 1 months
    "risk_label": ["very_high", "high", "medium", "low"],
    "risk_threshold": [0.9, 0.75, 0.4],
}
```

```python
from src.calculation_features import label_risk_level_category
```

```python
df_mock["MCC"] = df_mock["MCC"].fillna("__missing__")
```

```python
df_mock = label_risk_level_category(
    df=df_mock, datetime_col="Transaction Datetime", config=dynamic_high_risk_mcc_config
)
```

```python
df_mock[df_mock["PANNumber"] == 271191714][
    # df_mock[df_mock['MCCRiskLevel'].isna()][
    [
        "PANNumber",
        "Transaction Serial No",
        "Transaction Datetime",
        "MCC",
        "MCCRiskLevel",
    ]
].sort_values(by="Transaction Datetime", ascending=True)
```

```python
# Define the reference window
max_date = df_mock["Transaction Datetime"].max()
min_date = max_date - pd.DateOffset(months=12)

# Filter transactions in the reference window
reference_df = df_mock[
    (df_mock["Transaction Datetime"] >= min_date)
    & (df_mock["Transaction Datetime"] <= max_date)
]

# Group by MCC and normalized
agg_df = reference_df.groupby("MCC")["Confirmed"].agg("sum")
```
