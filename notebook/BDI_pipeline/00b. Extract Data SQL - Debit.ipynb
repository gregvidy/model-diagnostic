# Extract Triggered Rules Data

```python
df_dc_ml = pd.read_parquet("data/ready_to_train/df_debit_final_25_feats_20250702.parquet")
```

```python
from src.model_pipeline import ModelPipeline
pipeline = ModelPipeline()
```

```python
df_splits = pipeline.split_data_by_date(
    df=df_dc_ml,
    date_column="Transaction Datetime",
    split_date="2025-05-01"
)
```

```python
query_rules = read_query_file("sql/join_data/02a_debit_card_triggered_rules.sql")
query_rules = query_rules.format(
    start_date="2025-05-01",
    end_date="2025-05-26"
)
```

```python
df_debit_rules = pd.read_sql(query_rules, conn)
```

```python
df_eval = df_debit_rules[df_debit_rules['Transaction Serial No'].isin(df_splits['df_test'].index)]
del df_debit_rules
```

```python
def check_triggered_rules(val):
    return 1 if pd.notnull(val) and str(val).strip() != "" else 0
```

```python
df_eval['IsTriggeredRules'] = df_eval['Triggered_Rules'].apply(check_triggered_rules)
df_eval['Confirmed'] = df_eval['Confirmed'].fillna(0)
df_eval['Confirmed'] = df_eval['Confirmed'].astype(int)
```

## Calculation of Triggered Rules Effectiveness

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix
```

```python
y_pred = df_eval['IsTriggeredRules']
y_true = df_eval['Confirmed']
cm = confusion_matrix(y_true, y_pred)

# plot using seaborn heatmap
tn, fp, fn, tp = cm.ravel()
labels = ["Non-Fraud", "Fraud"]
plt.figure(figsize=(6,4))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
            xticklabels=labels, yticklabels=labels)
plt.xlabel("Predicted by Rules")
plt.ylabel("Actual")
plt.title("Confusion Matrix")
plt.tight_layout()
plt.show()
print(f"Recall/True Positive Rate: The rules got {np.round(tp/(tp+fn)*100, 2)}% of correctly predicted fraud from all of the actual fraud transactions")
print(f"Precision: The rules got {np.round(tp/(tp+fp)*100, 2)}% of correctly predicted fraud from all transactions that predicted as fraud")
print(f"Alert Rate: The rules will bring {np.round(((tp+fp)/(fn+tp+fp+tn))*100, 2)}% of the records from total to the Alert rate")
```
